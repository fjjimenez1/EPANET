# request | response


```python
import requests

url = "https://api.deepseek.com/chat/completions"
api_key = "sk-20d206a6d7f5484482f5185e7861478d"

headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {api_key}"
}

data = {
    "model": "deepseek-chat",
    "messages": [
        {"role": "system", "content": "eres un cocinero especialisat en comida ecuatoriana."},
        {"role": "user", "content": "como hacer un yapingacho"}
    ],
    "stream": False
}

response = requests.post(url, headers=headers, json=data)

if response.status_code == 200:
    respuesta = response.json()
else:
    print(f"Error: {response.status_code}")
    print(response.text)
```

# Analisis de una página web usando deepseek

```python
import requests

url = "https://www.masappec.com"
response = requests.get(url)

if response.status_code == 200:
    page_content = response.text
else:
    print(f"Failed to fetch the web page. Status code: {response.status_code}")
    exit(1)
```

```python
from openai import OpenAI

client = OpenAI(api_key="sk-20d206a6d7f5484482f5185e7861478d", base_url="https://api.deepseek.com")

prompt = f"<content>{page_content}</content>Por favor realiza un resumen conciso y claro del contenido de la página web."

response = client.chat.completions.create(
    model="deepseek-chat",
    messages=[
        {"role": "system", "content": "eres un experto analizando taxonomía de páginas webs"},
        {"role": "user", "content": prompt},
    ],
    stream=False
)

print(response.model_dump_json())
```

```python
from IPython.display import display, Markdown

display(Markdown(response.choices[0].message.content))
```
